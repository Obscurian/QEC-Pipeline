```python
def print_basis(basis):
    if not basis:
        print("{ }")
        return

    # Check the vector type using your existing VEC_TYPE enum
    is_row = basis[0].type == VEC_TYPE.ROW
    
    if is_row:
        # Format for Row Vectors: { [ 0 0 1], [1 0 1], [0 1 1] }
        vec_strs = []
        for v in basis:
            vals = " ".join(str(x) for x in v._vec)
            vec_strs.append(f"[{vals}]")
        print("{ " + ", ".join(vec_strs) + " }")
    
    else:
        # Format for Column Vectors:
        # { [0], [1], [0] }
        # { [0], [0], [1] }
        # { [1], [1], [1] }
        dim = basis[0].DIM
        lines = []
        for i in range(dim):
            # Extract the i-th element of every vector and wrap it in brackets
            row_vals = ", ".join(f"[{v._vec[i]}]" for v in basis)
            lines.append(f"{{ {row_vals} }}")
            
        print("\n".join(lines))
```

```python
M = Z2_matrix.create([[1,0,1,1],[0,1,0,1],[0,0,1,1]])

print(M)
print()
print(M.rref)
print()
print_basis(M.row_space)
print()
print_basis(M.column_space)
print()
print_basis(M.kernel)
```
